```python
import threading

from fastapi import FastAPI

from gerador_transacoes import iniciar_simulacao
```

```python
app = FastAPI()
```

```python
transacoes = []
```

```python
@app.on_event("startup")
def iniciar():

    thread = threading.Thread(target=iniciar_simulacao, args=(transacoes,))

    thread.start()
```

```python
@app.get("/transacoes")
def get_transacoes():

    return transacoes
```

```python
@app.get("/ultima-transacao")
def get_ultima_transacao():

    return transacoes[-1]
```
